```python
from galtea import Galtea
```

```python
API_KEY = "<your_api_key_here>"
# Initialize the Galtea client with your API key
# Replace <your_api_key_here> with your actual API key
galtea = Galtea(api_key=API_KEY)
```

```python
# Load the product
# Replace "Test Product" with the actual name of your product
product = galtea.products.get_by_name("Test Product")
# Replace "cat" with the actual name of your test
test = galtea.tests.get_by_name(product.id, "cat")
metric_accuracy = galtea.metrics.get_by_name('accuracy-basic')
# Replace "v1.1" with the actual version name
version = galtea.versions.get_by_name(product.id, "v1.1")
# Create a new evaluation
evaluation = galtea.evaluations.create(test_id=test.id, version_id=version.id)
```

```python
test_cases = galtea.test_cases.list(test.id)
```

```python
from rag.run_rag import query_rag
from datetime import datetime
```

```python
# Launch evaluations with direct calls to the function
evaluation_tasks=[]
for test_case in test_cases:
    timeBeforeCall = datetime.now()
    model_answer = query_rag(test_case.input)
    timeAfterCall = datetime.now()
    evaluation_tasks.append(galtea.evaluation_tasks.create(
      metrics=[metric_accuracy.name],
      evaluation_id=evaluation.id,
      actual_output=model_answer,
      test_case_id=test_case.id,
      latency=(timeAfterCall - timeBeforeCall).total_seconds() * 1000
    ))
```

```python
# The wrapper to launch evaluations from the restful API
import requests
import json

def ask_rag(message):
    url = 'http://127.0.0.1:8000/v1/chat'
    headers = {
        'accept': 'application/json',
        'Content-Type': 'application/json'
    }
    data = {
        'message': message
    }
    
    response = requests.post(url, headers=headers, data=json.dumps(data))
    
    if response.status_code == 200:
        return response.json()  # Return the JSON response if the request was successful
    else:
        response.raise_for_status()  # Raise an error for bad responses
```

```python
# Launch evaluations with calls to the API endpoint
evaluation_tasks=[]
for test_case in test_cases:
    timeBeforeCall = datetime.now()
    model_answer = ask_rag(test_case.input)["reply"]
    timeAfterCall = datetime.now()
    evaluation_tasks.append(galtea.evaluation_tasks.create(
      metrics=[metric_accuracy.name],
      evaluation_id=evaluation.id,
      actual_output=model_answer,
      test_case_id=test_case.id,
      latency=(timeAfterCall - timeBeforeCall).total_seconds() * 1000
    ))
```
